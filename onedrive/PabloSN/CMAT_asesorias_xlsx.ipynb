```python
import pandas as pd

path = input("Introduce el path del archivo Excel (copia y pega, como en las instrucciones): ").strip()

try:
    df = pd.read_excel(path)
    print("Archivo cargado correctamente")

except FileNotFoundError:
    print("Error: No se encontró el archivo en la ruta proporcionada.")

except ValueError:

    print("Error: El archivo no parece ser un archivo Excel válido.")

except Exception as e:
    print(f"Ocurrió un error al cargar el archivo: {e}")
```

```python
def column_otros(df):

  re_name = {

      "Hora de inicio": "fecha",
      "Escanear el ID del estudiante": "id",
      "CARRERA": "carrera",
      "Indicar el semestre que actualmente cursa el estudiante": "semestre",
      # "Elija su nombre como profesor que imparte la asesoría": "profesor",
      "Elija la materia de la asesoría que imparte": "materia"
  }


  df = df.rename(columns = re_name)
  idx_materia = df.columns.get_loc("materia")
  concat_df_temas = df.columns[idx_materia + 1:]

  df["tema"] = df[concat_df_temas].apply(lambda fila: " ".join(fila.dropna().astype(str)), axis=1)
  df = df.drop(columns = concat_df_temas)

  interest_columns = ["fecha", "id", "carrera", "semestre", "profesor", "materia", "tema"]
  df = df[[col for col in interest_columns if col in df.columns]]
  df["periodo"] = "P" # CAMBIALO DEPENDIENDO SI ES P: Primavera o O: otoño

  # df.to_excel("clean_excel.xlsx", index = False)
  return df


df = column_otros(df)
df
```
