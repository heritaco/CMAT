# Se explica como se crea el modulo que concatena los años del concentrado de CMAT

El resultado es un modulo en 'scripts/rexcel.py' de read excel.

```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

# Reading the dataframes for each year

```python
years = [19, 20, 21, 22, 23, 24]

# create a dictionany to hold dataframes
# data\concentrado_base_CMAT.xlsx
df = {}
for year in years:
    df[year] = pd.read_excel('data/onedrive/concentrado_base_CMAT.xlsx', sheet_name=f'20{year}')
    df[year]['Year'] = 2000 + year
```

```python
df[19].head()
```

```python
df[24].head()
```

```python
frames = [df[y] for y in years if y in df]
out = pd.concat(frames, ignore_index=True)
del frames, df  # if you’re done with them
```
